```python
import pandas as pd
ruta=("C:/Acintesa/Amjy_Documentos/Documentos_Amjy/amjy_personal/Practicas/Vcode/onlinef.csv")

dt=pd.read_csv(ruta)
```

```python
#información general  del dataset
print(dt.info())
print(dt.shape)
print(dt.columns)
```

```python
#Revision de columna 
#print(dt[dt["Unnamed: 13"]!="No"])

#columna no relevante, dato no indicado en el data set: eliminar
dt= dt.drop(columns= ["Unnamed: 13"])
print(dt.columns)

```

```python
#Resumen estadistico
dt.describe()
```

```python
# Columna Output: Current status of the order (e.g., pending, confirmed, delivered)- But, instead its has values like yes or no / It's not relevant, So  I decided to eliminate it
revision= dt[dt["Output"] != "Yes"]
#print(revision)

dt= dt.drop(columns=["Output"])  

print(dt.columns)
```

```python

opiniones= dt["Feedback"].value_counts()

print(opiniones)
```

```python
#feedback positivo por edad
Feedback= dt[dt["Feedback"] == "Positive"].groupby("Age").size().reset_index(name="Feedback positivo") #reset_index para indicar el nombre de la columna

Fagrupado=Feedback.sort_values(by="Feedback positivo",ascending=False)

print(Fagrupado)

```

```python
#Rango de edad del consumidor 

rango=[18,20,25,33]
etiqueta=["Entre 18 a 20", "Entre 20-25","mayor a 25"]

dt["Categoria"] = pd.cut(dt["Age"],bins=rango, labels=etiqueta)
resumen= dt["Categoria"].value_counts()

print(resumen)

```

```python
#Age and monthly income                         nota: sort_index=agrupa por el gruop by que te da el indice - sort_values por la cantidad que es el resultado

Income=dt["Monthly Income"].unique()
#print(Income)

#El rango de edad que mas consume sus ingresos y ocupacion

ingreso=dt[
    (dt["Categoria"]=="Entre 20-25") &
    (dt["Feedback"]=="Positive")
].groupby(["Monthly Income", "Occupation"]).size().sort_values(ascending=False)


print(ingreso)
```
